# Misturando Python com Manim

```python
%%manim -qm -v WARNING AreaInteligente

from manim import *

class AreaInteligente(Scene):
    def construct(self):
        titulo = Tex(r"Área de um quadrado")

        self.play(Write(titulo))

        self.play(titulo.animate.to_edge(UP,buff=0.1))

        self.wait()

        quadrado = Square(
            color=BLUE,
            fill_color=BLUE,
            fill_opacity=0.3,
            side_length=4
            )

        latex_n1 = MathTex(r"n",font_size=80)
        latex_n2 = MathTex(r"n",font_size=80)

        latex_n1.next_to(quadrado,RIGHT)
        latex_n2.next_to(quadrado,UP)

        self.play(Write(quadrado))

        self.play(Write(latex_n1),Write(latex_n2))

        grupo = VGroup(quadrado,latex_n1,latex_n2)

        self.play(grupo.animate.shift(LEFT*4))

        latex_area = MathTex(r"A=n^2",font_size=80)
        latex_A = MathTex(r"A",font_size=120) 

        latex_area.shift(RIGHT*3 + UP*2)

        self.play(Write(latex_area))

        self.wait()

        latex_A.move_to(quadrado.get_center())
        latex_A.set_z_index(1)

        self.play(Write(latex_A))

        # Definição do lado e area
        n = 2
        def area(lado):
            return lado ** 2

        valor_n = MathTex(r"n=",str(n),font_size=80)
        conta1 = MathTex(r"A=",str(n),r"^2",font_size=80)
        conta2 = MathTex(r"A=",str(area(n)),font_size=80)

        grupo_conta = VGroup(conta1,conta2).arrange(DOWN)

        valor_n.next_to(latex_area,DOWN,buff=1)
        valor_n.shift(LEFT*0.5)

        grupo_conta.next_to(valor_n,DOWN,buff=0.5)
        grupo_conta.shift(RIGHT*0.5)

        self.play(Write(valor_n))

        self.wait()

        self.play(Write(conta1))

        self.wait()

        self.play(Write(conta2))

        self.wait()
```
