# Animações Inteligentes

Nosso projeto para mostrar o cálculo da área de um quadrado parece bom, porem o que acontece se quisermos mudar o valor de $n$?

Nesse caso podemos alterar todas as linhas que contem o LaTeX que contem de algumas forma o valor de $n$, e essa é uma solução perfeitamente valida, mas isso pode acabar se tornando um problema rapidamente se nossa animação, por exemplo, conter diversas linhas para o desenvolvimento de uma conta.

## Então como podemos resolver esse problema?

Nesse caso podemos usar uma variável para armazenar o valor de $n$ e usar uma função para calcular o valor da área. Vamos ver um exemplo simples:

````python
n = 2

def area(lado):
    return lado ** 2

print(area(n))
````

Até agora estamos usando variáveis do Python para armazenar objetos Manim, agora estamos usando uma variável `n` para armazenar um valor, nesse caso `2`.

Na linha abaixo definimos uma função chamada `area()` essa função recebe um valor `lado` e retorna o lado elevado ao quadrado.

## Mas como podemos usar isso na nossa animação?

Para usar essa estrutura na nossa animação podemos fazer da seguinte forma:

````python
%%manim -qm -v WARNING AreaInteligente

from manim import *

class AreaInteligente(Scene):
    def construct(self):
        titulo = Tex(r"Área de um quadrado")

        self.play(Write(titulo))

        self.play(titulo.animate.to_edge(UP,buff=0.1))

        self.wait()

        quadrado = Square(
            color=BLUE,
            fill_color=BLUE,
            fill_opacity=0.3,
            side_length=4
            )

        latex_n1 = MathTex(r"n",font_size=80)
        latex_n2 = MathTex(r"n",font_size=80)

        latex_n1.next_to(quadrado,RIGHT)
        latex_n2.next_to(quadrado,UP)

        self.play(Write(quadrado))

        self.play(Write(latex_n1),Write(latex_n2))

        grupo = VGroup(quadrado,latex_n1,latex_n2)

        self.play(grupo.animate.shift(LEFT*4))

        latex_area = MathTex(r"A=n^2",font_size=80)
        latex_A = MathTex(r"A",font_size=120) 

        latex_area.shift(RIGHT*3 + UP*2)

        self.play(Write(latex_area))

        self.wait()

        latex_A.move_to(quadrado.get_center())
        latex_A.set_z_index(1)

        self.play(Write(latex_A))

        # Definição do lado e area
        n = 2
        def area(lado):
            return lado ** 2

        valor_n = MathTex(r"n=",str(n),font_size=80)
        conta1 = MathTex(r"A=",str(n),r"^2",font_size=80)
        conta2 = MathTex(r"A=",str(area(n)),font_size=80)

        grupo_conta = VGroup(conta1,conta2).arrange(DOWN)

        valor_n.next_to(latex_area,DOWN,buff=1)
        valor_n.shift(LEFT*0.5)

        grupo_conta.next_to(valor_n,DOWN,buff=0.5)
        grupo_conta.shift(RIGHT*0.5)

        self.play(Write(valor_n))

        self.wait()

        self.play(Write(conta1))

        self.wait()

        self.play(Write(conta2))

        self.wait()
````

Vamos ver a linhas que alteramos para entender o que está acontecendo. Primeiro fizemos a inserção da estrutura que discutimos anteriormente:
```python
n = 2

def area(lado):
	return lado ** 2
```

Logo abaixo fizemos uma alteração nas definições dos objetos `MathTex()`:
```python
valor_n = MathTex(r"n=",str(n),font_size=80)
conta1 = MathTex(r"A=",str(n),r"^2",font_size=80)
conta2 = MathTex(r"A=",str(area(n)),font_size=80)
```

Perceba que tudo que fizemos foi alterar o texto de entrada para o LaTeX, porem, estamos fazendo uso da variável `n` e da função `area()` que definimos anteriormente, vamos ver o que esta acontecendo de fato nessas linhas, vamos tomar de exemplo a primeira linha:
```python
valor_n = MathTex(r"n=",str(n),font_size=80)
```

Aqui podemos ver que definimos uma variável chamada `valor_n`, essa variável armazena o objeto Manim `MathTex()`, como vimos anteriormente, o objeto `MathTex()` precisa de uma `string` (ou um texto) para definir o que será escrito, nesse caso estamos passando duas `strings`, a primeira é `r"n="` e a segunda fazemos uso da função `str()` para transformar `n`, que atualmente é um número, em uma `string`, nesse caso, o objeto `MathTex()` irá juntar as duas strings em uma, como definimos n=2 vamos ter que a string final para esse objeto `MathTex()` será: `r"n=2"` que é justamente o que queremos.

Os outros dois objetos `MathTex()` seguem o mesmo principio:
```python
conta1 = MathTex(r"A=",str(n),r"^2",font_size=80)
conta2 = MathTex(r"A=",str(area(n)),font_size=80)
```

Nesse caso temos que `conta1` terá o LaTeX correspondente a: `r"A=2^2"` e `conta2` terá o LaTeX correspondente a: `r"A=4^2"`.

Tente alterar o valor de n para 5, por exemplo, e renderize a animação para ver o resultado. Tente colocar outros valores arbitrários e veja o que acontece.
## Por que isso é relevante?
O grande ponto de se usar uma linguagem de programação para se criar animações é poder usar toda a lógica de programação e ferramentas dessas linguagens para criar animações que sejam inteligentes, ou seja, animações que fazem alterações seguindo regras específicas sem que o animador (ou nesse caso, o professor) se preocupe em checar todos os objetos da sua animação.

Nesse caso fizemos o uso de uma variável e uma função do Python para criar uma animação extremamente fácil de ser adaptada para valores diferentes de $n$.

Vamos mudar o valor de `n` no código para 8 por exemplo:

````python
%%manim -qm -v WARNING AreaInteligente2

from manim import *

class AreaInteligente2(Scene):
    def construct(self):
        titulo = Tex(r"Área de um quadrado")

        self.play(Write(titulo))

        self.play(titulo.animate.to_edge(UP,buff=0.1))

        self.wait()

        quadrado = Square(
            color=BLUE,
            fill_color=BLUE,
            fill_opacity=0.3,
            side_length=4
            )

        latex_n1 = MathTex(r"n",font_size=80)
        latex_n2 = MathTex(r"n",font_size=80)

        latex_n1.next_to(quadrado,RIGHT)
        latex_n2.next_to(quadrado,UP)

        self.play(Write(quadrado))

        self.play(Write(latex_n1),Write(latex_n2))

        grupo = VGroup(quadrado,latex_n1,latex_n2)

        self.play(grupo.animate.shift(LEFT*4))

        latex_area = MathTex(r"A=n^2",font_size=80)
        latex_A = MathTex(r"A",font_size=120) 

        latex_area.shift(RIGHT*3 + UP*2)

        self.play(Write(latex_area))

        self.wait()

        latex_A.move_to(quadrado.get_center())
        latex_A.set_z_index(1)

        self.play(Write(latex_A))

        # Definição do lado e area
        n = 8 
        def area(lado):
            return lado ** 2

        valor_n = MathTex(r"n=",str(n),font_size=80)
        conta1 = MathTex(r"A=",str(n),r"^2",font_size=80)
        conta2 = MathTex(r"A=",str(area(n)),font_size=80)

        grupo_conta = VGroup(conta1,conta2).arrange(DOWN)

        valor_n.next_to(latex_area,DOWN,buff=1)
        valor_n.shift(LEFT*0.5)

        grupo_conta.next_to(valor_n,DOWN,buff=0.5)
        grupo_conta.shift(RIGHT*0.5)

        self.play(Write(valor_n))

        self.wait()

        self.play(Write(conta1))

        self.wait()

        self.play(Write(conta2))

        self.wait()
````
