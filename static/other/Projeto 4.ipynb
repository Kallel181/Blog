# Graficos com Manim
## Planos cartesianos

```python
%%manim -qm -v WARNING grafico1

from manim import *

class grafico1(Scene):
    def construct(self):
        plano_catesiano = Axes(
            x_range=[-2, 2, 1],
            y_range=[-0.5, 4, 1],
            x_length=10,
            axis_config={"color": WHITE},
        )
        
        legenda_plano_cartesiano = plano_catesiano.get_axis_labels()

        def func(x):
            return x**2  

        curva = plano_catesiano.plot(func,color=BLUE)  

        self.add(plano_catesiano,legenda_plano_cartesiano,curva)
```

# Value Trackers 
## Exemplo 1

```python
%%manim -qm -v WARNING ExemploValueTracker

from manim import *

class ExemploValueTracker(Scene):
    def construct(self):
        linha_numerica = NumberLine()
        #linha_numerica.add_numbers()
        
        seta = Vector(DOWN)
        rotulo = MathTex("x")

        seta.next_to(linha_numerica.n2p(0),UP)
        rotulo.next_to(seta,UP)

        tracker = ValueTracker(0)

        def update_pointer(objeto):
            posicao = linha_numerica.n2p(tracker.get_value())
            objeto.next_to(posicao, UP)
            rotulo.next_to(objeto, UP)

        seta.add_updater(update_pointer)

        self.add(linha_numerica, seta, rotulo)        

        self.play(tracker.animate.set_value(5))
        self.wait(0.5)

        self.play(tracker.animate.set_value(3))
        self.wait(0.5)
        
        self.play(tracker.animate.increment_value(-2))
        self.wait(0.5)

        self.play(tracker.animate.set_value(-5))
        self.wait(0.5)

        self.play(tracker.animate.set_value(0))
        self.wait(0.5)
```

# O que a n2p() faz?

```python
%%manim -qm -v WARNING ComparandoCoordenadasReta

from manim import *

class ComparandoCoordenadasReta(Scene):
    def construct(self):
        # Criamos a reta fora do centro para evidenciar a diferença
        linha = NumberLine(x_range=[-3, 3, 1], length=6).shift(RIGHT * 1.5 + DOWN * 1)
        linha.add_numbers()

        num_reta = 2
        # n2p converte o valor da reta (2) na coordenada global do Manim [x, y, z]
        ponto_manim = linha.n2p(num_reta)

        dot = Dot(ponto_manim, color=RED)

        # Rótulo mostrando o valor na reta
        texto_reta = Text(f"Número na Reta: {num_reta}", color=YELLOW, font_size=24)
        texto_reta.to_corner(UL)

        # Rótulo mostrando as coordenadas reais no Manim (Corrigido: aligned_edge=LEFT)
        x_m, y_m, _ = ponto_manim
        texto_manim = Text(
            f"Coordenadas no Manim: [{x_m:.2f}, {y_m:.2f}, 0.00]",
            color=RED,
            font_size=24,
        ).next_to(texto_reta, DOWN, aligned_edge=LEFT)

        self.add(linha, dot, texto_reta, texto_manim)
```

## Exemplo 2

```python
%%manim -qh -v WARNING GraficoValeuTracker

from manim import *

class GraficoValeuTracker(Scene):
    def construct(self):
        plano_catesiano = Axes(
            x_range=[-2, 2, 1],
            y_range=[-0.5, 4, 1],
            x_length=10,
            axis_config={"color": WHITE, "include_numbers": True},
        )

        legenda_plano_cartesiano = plano_catesiano.get_axis_labels()

        def func(x):
            return x**2

        curva = plano_catesiano.plot(func, color=BLUE)

        x = ValueTracker(0)

        ponto = Dot(
            point=plano_catesiano.c2p(x.get_value(), func(x.get_value()))
        )

        def update_ponto(objeto):
            novo_ponto = Dot(
                point=plano_catesiano.c2p(x.get_value(), func(x.get_value()))
            )
            objeto.become(novo_ponto)

        ponto.add_updater(update_ponto)

        self.add(plano_catesiano, legenda_plano_cartesiano, curva, ponto)
        self.wait(0.5)

        self.play(x.animate.set_value(2))
        self.wait(0.5)

        self.play(x.animate.set_value(-2))
        self.wait(0.5)

        self.play(x.animate.set_value(0))
        self.wait(0.5)
```

# O que o c2p() faz?

```python
%%manim -qh -v WARNING CoordsToPointExample

from manim import *

class CoordsToPointExample(Scene):
    def construct(self):
        ax = Axes().add_coordinates()

        # a dot with respect to the axes
        dot_axes = Dot(ax.coords_to_point(2, 2), color=GREEN)
        lines = ax.get_lines_to_point(ax.c2p(2,2))

        # a dot with respect to the scene
        # the default plane corresponds to the coordinates of the scene.
        plane = NumberPlane()
        dot_scene = Dot((2,2,0), color=RED)

        self.add(plane, dot_scene, ax, dot_axes, lines)
```
