````python
%%manim -qm -v WARNING PrimeiraCena

from manim import *

class PrimeiraCena(Scene):
    def construct(self):
        # Exemplo basico de animacao Manim
        circulo = Circle(color=BLUE)
        quadrado = Square(color=RED)
        
        self.play(Create(circulo))
        
        self.play(Transform(circulo, quadrado))

        self.wait()

        self.play(FadeOut(circulo))
        
        title = Tex(r"Isso e um texto \LaTeX")
        basel = MathTex(r"\sum_{n=1}^\infty \frac{1}{n^2} = \frac{\pi^2}{6}")

        VGroup(title, basel).arrange(DOWN)

        self.play(
            Write(title),
            FadeIn(basel, shift=DOWN),
        )

        self.wait()
````

A primeira linha é um `import`, isso sinaliza para o Python que vamos usar uma biblioteca que já possui funcionalidades prontas, ou seja, toda vez que criamos um quadrado, escrevemos um texto LaTeX ou usamos um método de animação, essas funcionalidades veem todas dessa biblioteca.

Seria perfeitamente possível acessar esse código para descobrir todos os objetos e animações que o Manim nos fornece, porem, olhar para o código-fonte diretamente para encontrar funcionalidades específicas pode ser (e acredite em mim, realmente vai ser) uma tarefa tediosa e trabalhosa, por isso, sempre que estivermos em dúvida sobre alguma funcionalidade ou quisermos descobrir o que mais há na ferramenta nós podemos olhar a [documentação](https://docs.manim.community/en/stable/index.html). 

Na documentação são só temos acesso a todas as funcionalidades da ferramenta, mas também temos acesso a [guia de instalação](https://docs.manim.community/en/stable/installation.html) e a uma [galeria de exemplos](https://docs.manim.community/en/stable/examples.html).

As próximas duas linhas são definições de uma `class` e uma função usando `def`, existe um porquê dessas linhas existirem ali, porem para entendermos porque essas linhas estão ali precisamos mergulhar mais a fundo no funcionamento da ferramenta, e isso não seria muito produtivo. Algo que podemos fazer é usar o **método da caixa preta**, existe um porquê daquelas linhas, porem, vamos assumir que o Manim funciona assim e que essas linhas são necessárias sem nos preocuparmos no que essas linhas serão usadas e, porque elas existem.

Mais a baixo temos o corpo da nossa animação:
```python
circulo = Circle(color=BLUE)
quadrado = Square(color=RED)

self.play(Create(circulo))
self.play(Transform(circulo, quadrado))
self.wait()
self.play(FadeOut(circulo))

title = Tex(r"Isso e um texto \LaTeX")
basel = MathTex(r"\sum_{n=1}^\infty \frac{1}{n^2} = \frac{\pi^2}{6}")

self.play(
    Write(title),

```

Vamos linha a linha para entender o que está acontecendo. Duas primeiras linhas definem objeto Manim:
```python
circulo = Circle(color=BLUE)
quadrado = Square(color=RED)
```

Nesse caso estamos criando duas variáveis, uma chamada `circulo` e outra chamada `quadrado`, nessa variáveis estamos guardando objetos Manim que são criados usando as funções `Circle()` e `Square()` que como o nome sugere, são usados para criar círculos e quadrados respectivamente.

Dentro dessas funções estamos passando um atributo, nesse caso o atributo `color`, o círculo recebe a cor `BLUE` e o quadrado a cor `RED`. Já existe uma série de cores predefinidas dentro do Manim, na [documentação](https://docs.manim.community/en/stable/reference/manim.utils.color.manim_colors.html) podemos encontrar uma tabela com todas as cores disponíveis.

Toda criação de um objeto Manim segue esse padrão, vamos criar uma variável para armazenar o objeto, vamos usar uma função para criar o objeto e vamos passar atributos específicos para customizar os objetos.

As próximas quatro linhas é onde definimos a animação de fato:
```python
self.play(Create(circulo))
self.play(Transform(circulo, quadrado))
self.wait()
self.play(FadeOut(circulo))
```

Todo método de animação é passado em um `self.play()`, novamente, isso tem um porquê, porem não vamos nos preocupar com isso agora, vamos assumir que isso precisa ser feito e seguir em frente. A primeira animação que usamos é a `Create()` perceba que criar um objeto Manim, ou seja, definir uma variável para armazenar um objeto Manim não significa que esse objeto está em cena, para colocar algo em cena precisamos especificar que algo deve ser mostrado.

Vamos analisar mais de perto, nas primeiras linhas definimos um quadrado e um círculo, porem, se observamos no começo da animação não temos os dois em cena, o primeiro a aparece é o círculo, o círculo só aparece quando definimos uma forma para que ele possa aparecer, nesse caso usamos o método `Create()` para indicar que o círculo deve ser criado dentro da cena.

O método `Create()` possui uma forma característica de anima objetos, e existem muitas outras formas de se colocar objetos em cena, como, por exemplo, `FadeIn()`, `Write()`, `GrowFromCenter()` ...

Logo em seguida usamos o método `Transform(A,B)` esse método sempre recebe ao menos dois atributos, que nesse caso, são dois objetos Manim, e ele faz o que o nome sugere, ele vai transformas o objeto A no objeto B, por isso que logo após o círculo ser criado na tela, ele é transformado em um quadrado.

Mais a frente, o método `self.wait()` é inserido, esse método adiciona uma pausa na animação, por padrão essa pausa é de um segundo, porem podemos especificar a pausa sem segundo dentro dos parênteses com, por exemplo: 
`self.wait(2)` para esperar 2 segundos ou `self.wait(20)` para esperar 20 segundos ou ainda `self.wait(0.5)` para esperar meio segundo.

Por último, usamos o método `FadeOut()` para remover a figura da tela, porem, se está prestando atenção talvez tenha percebido algo estranho, na tela temos o `quadrado` e não o `circulo` **porque escrevemos `FadeOut(circulo)` e não `FadeOut(quadrado)` ?**

A reposta é simples, quando usamos o método `self.play(Transform(circulo, quadrado))` nós não substituirmos o círculo pelo quadrado, nós transformamos o círculo em um quadrado. Por isso, para tirar o quadrado da tela, precisamos usar `FadeOut(circulo)`.

Se isso for um comportamento confuso (e acredite, normalmente é e gera muitos problemas e dor de cabeça), podemos usar o `self.play(ReplacementTransform(circulo, quadrado))`, dessa forma teremos a mesma animação, porem o quadrado tomara o lugar do círculo, e quando quisermos removê-lo de cena podemos usar o `FadeOut(quadrado)`.

O fim da animação temos uma última parte:
```python
title = Tex(r"Isso e um texto \LaTeX")
basel = MathTex(r"\sum_{n=1}^\infty \frac{1}{n^2} = \frac{\pi^2}{6}")

VGroup(title, basel).arrange(DOWN)

self.play(
    Write(title),
    FadeIn(basel, shift=DOWN),
)

self.wait()
```

Essa parte segue o mesmo padrão que estávamos usando anteriormente, as duas primeiras linhas definem dois objetos Manim:
```python
title = Tex(r"Isso e um texto \LaTeX")
basel = MathTex(r"\sum_{n=1}^\infty \frac{1}{n^2} = \frac{\pi^2}{6}")
```

Nesse caso usamos os métodos `Tex()` e `MathTex()`. O método `Tex()` é usado para escrever texto, e o método `MathTex()` é usado para escrever formulas. Como argumento dessas funções precisamos passar uma string com um texto LaTeX.

Uma string nada mais é que um nome alternativo para texto, nesse caso, uma string sempre é escrita entre aspas "", mas especificamente para o Python e especialmente para strings com comando LaTeX precisamos colocar um `r` na frente (novamente, isso tem um porquê, mas não vamos nos preocupar com isso agora).

O primeiro objeto definimos um texto comum, porem usamos o comando `\LaTeX`, que basicamente coloca a logo do LaTeX ajustada.

O segundo objeto é a série infinita de basel uma série famosa resolvida por Euler em 1735.

$$\sum_{n=1}^\infty \frac{1}{n^2} = \frac{\pi^2}{6}$$

Logo abaixo disso temos algo novo:
```python
VGroup(title, basel).arrange(DOWN)
```

`VGroup()` é uma função extremamente útil dentro do Manim, talvez uma das mais uteis, tudo que ela faz é agrupar objetos, porem existe uma série de funcionalidades associadas a esses grupos, uma delas é o `arrange()`.

Até agora fizemos a criação de dois objetos, `title` e `basel`, ambos são textos em LaTeX, porem por padrão o Manim cria todos os objeto no centro da tela, nesse caso, não seria interessante deixar dois textos sobrepostos (certamente não seria muito fácil ler eles), por isso precisamos arrumar os textos de forma que eles não fiquem sobrepostos.

Uma das maneiras de se fazer isso é usando a função `arrange()`, essa função recebe um atributo de direção, por padrão Manim possui uma série de vetores pré-configurados, segue alguns exemplos:

```python
ORIGIN: Vector3D = np.array((0.0, 0.0, 0.0))
"""The center of the coordinate system."""  

UP: Vector3D = np.array((0.0, 1.0, 0.0))
"""One unit step in the positive Y direction."""  

DOWN: Vector3D = np.array((0.0, -1.0, 0.0))
"""One unit step in the negative Y direction."""  

RIGHT: Vector3D = np.array((1.0, 0.0, 0.0))
"""One unit step in the positive X direction."""  

LEFT: Vector3D = np.array((-1.0, 0.0, 0.0))
"""One unit step in the negative X direction."""  

IN: Vector3D = np.array((0.0, 0.0, -1.0))
"""One unit step in the negative Z direction."""  

OUT: Vector3D = np.array((0.0, 0.0, 1.0))
"""One unit step in the positive Z direction."""
```

Nesse caso estamos passando para a função `arrange()` o atributo `DOWN`, dessa forma, todos os elementos do nosso `VGroup()` serão alinhados um abaixo do outro.

Perceba também que os elementos estarão dispostos na ordem em que aparecem dentro do `VGroup()`.

Poderíamos passar qualquer outra direção dentro do `arrange()` como `UP`, `LEFT` ou `RIGHT`:
```python
VGroup(title, basel).arrange(UP)
```

Ou ainda criar um vetor qualquer usando o `np.array()`:
```python
vetor = np.array((2.0, 1.0, 0.0))

VGroup(title, basel).arrange(vetor)
```

Ou ainda poderíamos fazer uma combinação linear entre os vetores padrão do Manim:
```python
VGroup(title, basel).arrange(2*RIGHT + UP)
#Essa combinação linear tem o mesmo efeito do vetor feito no exemplo acima
```

Por último na nossa animação temos:
```python
self.play(
    Write(title),
    FadeIn(basel, shift=DOWN),
)

self.wait()
```

Essa seção apresenta poucas coisas novas, porem tem alguns detalhes interessantes.

Primeiro, o método `self.play()` agora possui dois atributos, um `Write(title)` e `FadeIn(basel, shift=DOWN)`. Quando passamos mais de uma animação dentro do `self.play()` teremos que ambas as animações irão começar ao mesmo tempo, ou seja, elas irão acontecer em paralelo.

Essa é uma funcionalidade que precisa ser usadas com cautela dentro do Manim, uma vez que, ter coisas de mais acontecendo ao mesmo tempo, na tela pode ser distrativo.

Por último, a outra funcionalidade nova está na animação do basel, `FadeIn(basel, shift=DOWN)`, aqui usamos o método `FadeIn()` para criar nossa fórmula na tela, porem usamos um atributo extra chamado `shift`, esse atributo indica a direção da animação, nesse caso ao invés de simplesmente aparecer na tela, a fórmula irá aparecer na tela enquanto faz um leve movimento de cima para baixo.

A última linha da animação é novamente um `self.wait()`, que irá esperar 1 segundo antes de acabar a animação.
